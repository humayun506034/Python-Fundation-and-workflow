Operators in python

```python
x=10; y=3

sum=x+y #summation
sub=x-y #subtraction
mul=x*y #multiplication
div=x/y #division
mod=x%y #modulus

print("Sum:", sum, "Subtraction:", sub, "Multiplication:", mul, "Division:", div, "Modulus:", mod)
```

Comparison Operators

```python
x=10; y=3

greater_than=x>y 
greater_than_equal_to=x>=y
less_than=x<y 
less_than_equal_to=x<=y
equal_to=x==y 
not_equal_to=x!=y 

print("Greater than:", greater_than, "Less than:", less_than, "Equal to:", equal_to, "Not equal to:", not_equal_to)
```

Logical Operator

```python
x=10; y=5; z=3

result = (x>y) and (x>z) # logical AND
print("Result of logical AND:", result)
```

```python
x=10; y=3 ; z=15
result = (x>y) or (x>z) # logical OR
print("Result of logical OR:", result)
```
