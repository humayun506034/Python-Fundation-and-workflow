Conditionals statement in Python

if - else

```python
raining = False
if raining==True:
    print("Rain coat is necessary")
else:
    print("Go without a rain coat")
```

```python
raining = True
muddy = True
if raining == True:
    print("Rain coat is necessary")
    if muddy == True:
        print("Bring Sandle")
    else:
        print("Bring Snikers")
else:
    print("Go without a rain coat")
```

elif - if

```python
taka = 1000

if taka > 500:
    print("I will plan a journey")
elif taka > 400:
    print("I will order at home")
elif taka > 300:
    print("I will go foodcart")
else:
    print("I prefer savings")
```
