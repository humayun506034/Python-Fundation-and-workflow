Conditionals statement in Python

if - else

```python
raining = False
if raining==True:
    print("Rain coat is necessary")
else:
    print("Go without a rain coat")
```

nested if - else

```python
raining = True
muddy = True
if raining == True:
    print("Rain coat is necessary")
    if muddy == True:
        print("Bring Sandle")
    else:
        print("Bring Snikers")
else:
    print("Go without a rain coat")
```

elif - if

```python
taka = 1000

if taka > 500:
    print("I will plan a journey")
elif taka > 400:
    print("I will order at home")
elif taka > 300:
    print("I will go foodcart")
else:
    print("I prefer savings")
```

Loop in python

for loop

```python
for i in range(10):
    print(i, "Hello world")
```

```python
for i in range(1, 11, 2):
    print(i, "Hello world")
```

Nested for loop

```python
for i in range(1, 10):
    for j in range(1, 5):
        print(i, j)
```

While Loop in Python

```python
count = 1

while count < 10:
    print(count)
    # count=count+1
    count +=1
```

Break and Continue Statements in Python

```python
sum = 0
for i in range(1, 10):
    sum += i
    print(i, "is added to the sum")
print(sum)
```

Continue

```python
sum = 0
for i in range(1, 11):
    print(i, "is processing..")
    if i % 2 == 0:
        continue
    print(i, "is added to the sum")
    sum += i
print(sum)
```

Break

```python
accuracy = 95

for i in range(100):
    accuracy += 1
    print(accuracy)
    if accuracy == 100:
        break
```
