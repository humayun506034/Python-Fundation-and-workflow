Conditionals statement in Python

if - else

```python
raining = False
if raining==True:
    print("Rain coat is necessary")
else:
    print("Go without a rain coat")
```

nested if - else

```python
raining = True
muddy = True
if raining == True:
    print("Rain coat is necessary")
    if muddy == True:
        print("Bring Sandle")
    else:
        print("Bring Snikers")
else:
    print("Go without a rain coat")
```

elif - if

```python
taka = 1000

if taka > 500:
    print("I will plan a journey")
elif taka > 400:
    print("I will order at home")
elif taka > 300:
    print("I will go foodcart")
else:
    print("I prefer savings")
```

Loop in python

for loop

```python
for i in range(10):
    print(i, "Hello world")
```

```python
for i in range(1, 11, 2):
    print(i, "Hello world")
```

Nested for loop

```python
for i in range(1, 10):
    for j in range(1, 5):
        print(i, j)
```
