# **Taples**

```python
# Declaration

tuple_1 = (10,20,30)
# print(type(tupple_1), tupple_1)

# Decalare single value tuple
# first way
new_tuple=(10,)
# second way
second_tuple=tuple([10])
print(type(new_tuple), new_tuple)
print(type(second_tuple), second_tuple)

```

`<!-- List to tuple -->`

```python
list1 = [10,20,30]
tuple1=tuple(list1)
print(type(tuple1), tuple1)
```

**Access**

```python
tuple_1 = ( 10,20,30,10.5,"str",True )
# print(tuple_1)

# Indexing
print(tuple_1[4])
```

**Slicing**

```python
new_tuple = tuple_1[0:3]
print(new_tuple)
```

**Tuple Methods**

```python

tup = (10,20,30,40,10,10,20)

# count

print(tup.count(10))

# index

print(tup.index(10))
```

# **Set**

```python
# declaration

A = {1,2,3}
print(type(A), A)
```

```python
# empty structures

B = [] # empty list
print(type(B))

B = () # empty tuple
print(type(B))

S = set() # empty set
print(type(S))
```

```python
S = {1,2,3,4,1,2,5}

if ( 3 in S ):
  print("Present Here")
else:
  print("Not present")

sum = 0
for element in S:
  sum += element
print(sum)
```
