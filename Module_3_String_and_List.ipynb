# **Indexing And Slicing**

**Indexing**

```python
string= 'Hello world'
print(string[0])
```

**Slicing**

```python
text='Hello , I am Humayun'
# Call without step
first_word=text[0:5]
# call with step
first_word_with_two_step=text[0:5:2]

print(first_word)
print(first_word_with_two_step)
```

**Backword Indexing**

```python
print(first_word[-1])
```

# capitalize => প্রথম অক্ষর কে বড় হাতের করার *জন্য*

```python
my_name = "humayun kabir sobuj sobuj , sobuj"
print(my_name.capitalize())
```

# শুরুর দিকে কোন index থেকে substring টা শুরু হচ্ছে টা চেক করার জন্য

```python
find_index = my_name.find('sobuj')
print(find_index)
```

# লাস্ট দিকে কোন index থেকে substring টা শুরু হচ্ছে টা চেক করার জন্য ( Multple আছে )

```python
my_full_name= "humayun kabir sobuj sobuj , sobuj"

# প্রথম sobuj (বাম থেকে)
print(my_full_name.find('sobuj'))   # output: 14

# শেষ sobuj (ডান থেকে)
print(my_full_name.rfind('sobuj'))  # output: 28 (বা স্পেস অনুযায়ী শেষের s-এর ইনডেক্স)
```

# কোন word substring হিসেবে কয়বার আছে চেক করার জন্যঃ

```python
count_substring = my_full_name.count("sobuj")
print(count_substring)
```

# Replace Substring

```python
new_string = my_full_name.replace("sobuj","kabir")
print(my_full_name)
print(new_string)
```
