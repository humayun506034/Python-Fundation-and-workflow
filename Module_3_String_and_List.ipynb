# শুরুর দিকে কোন index থেকে substring টা শুরু হচ্ছে টা চেক করার জন্য

# লাস্ট দিকে কোন index থেকে substring টা শুরু হচ্ছে টা চেক করার জন্য ( Multple আছে )

```python
my_full_name= "humayun kabir sobuj sobuj , sobuj"

# প্রথম sobuj (বাম থেকে)
print(my_full_name.find('sobuj'))   # output: 14

# শেষ sobuj (ডান থেকে)
print(my_full_name.rfind('sobuj'))  # output: 28 (বা স্পেস অনুযায়ী শেষের s-এর ইনডেক্স)
```

# কোন word substring হিসেবে কয়বার আছে চেক করার জন্যঃ

```python
count_substring = my_full_name.count("sobuj")
print(count_substring)
```

# Replace Substring

```python
new_string = my_full_name.replace("sobuj","kabir")
print(my_full_name)
print(new_string)
```

# **List as Stack and Queue**

**sort method**

```python
second_numbers_array = [10,50,40,20,30,100,900,700]
second_numbers_array.sort()
print(second_numbers_array)

second_numbers_array.sort(reverse=True)
print(second_numbers_array)
```

**list as stack**

```python
stack = []

# Push on stack

stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

# Getting the top elements / Remove on stack

print(stack.pop())
print(stack.pop())
print(stack.pop())

print(f"Top element is {stack[-1]}")
```

**list as queue**

```python
queue = []

# Push on queue

queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

# Remove on queue

print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

print(f"Front element is {queue[0]}")

```

# **List Comprehension**

```python
even = []

# native way
for i in range(1,101):
  if i%2==0:
    even.append(i)
print(even)
```

```python
random_list = [ x for x in range(10) if x%2==0 ]
print(random_list)
```
